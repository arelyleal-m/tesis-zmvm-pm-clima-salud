```python
#USAR PARA PM2.5

import pandas as pd
import numpy as np
import os

# ========== CONFIGURACIÓN ==========
ruta_archivo = "AGREGAR RUTA DEL ARCHIVO"
ruta_salida = "AGREGAR RUTA DE CARPETA DE SALIDA"
nombre_archivo_salida = "AGREGAR NOMBRE DE SALIDA.csv”

# Cargar el archivo
df = pd.read_csv(ruta_archivo)

# Lista de columnas que representan los municipios
columnas_municipios = [
    'Acolman', 'Alvaro Obregon', 'Azcapotzalco', 'Benito Juarez', 'Coyoacan',
    'Cuajimalpa de Morelos', 'Cuauhtemoc', 'Ecatepec de Morelos', 'Gustavo A. Madero',
    'Iztapalapa', 'Miguel Hidalgo', 'Milpa Alta', 'Naucalpan de Juarez', 'Nezahualcoyotl',
    'Ocoyoacac', 'Tepotzotlan', 'Texcoco', 'Tlalnepantla de Baz', 'Tlalpan', 'Venustiano Carranza'
]

# Eliminar la columna "HORA" si existe
if "HORA" in df.columns:
    df.drop(columns=["HORA"], inplace=True)

# Agrupar cada 24 filas para formar días
df["grupo"] = df.index // 24

# Redondeo según norma: decimal entre 0 y 0.4 → piso; mayor a 0.4 → techo
def redondeo_personalizado(valor):
    if pd.isna(valor):
        return np.nan
    primer_decimal = int((valor * 10) % 10)
    return np.floor(valor) if primer_decimal <= 4 else np.ceil(valor)

    
# Función para calcular promedio por grupo (día)
def calcular_promedio(grupo):
    resultados = {}
    for col in columnas_municipios:
        datos = pd.to_numeric(grupo[col], errors='coerce')
        datos_validos = datos.dropna()
        if len(datos_validos) >= 18:
            promedio = datos_validos.sum() / len(datos_validos)
            resultados[col] = redondeo_personalizado(promedio)
        else:
            resultados[col] = np.nan
    return pd.Series(resultados)

# Aplicar la función de promedio por grupo
promedios = df.groupby("grupo").apply(calcular_promedio).reset_index(drop=True)

# Extraer la columna de fechas (una por día)
fechas_diarias = df["FECHA"].iloc[::24].reset_index(drop=True)
promedios.insert(0, "FECHA", fechas_diarias)

# Asegurar que columnas estén en orden alfabético después de "FECHA"
columnas_ordenadas = ["FECHA"] + sorted([col for col in columnas_municipios])
promedios = promedios[columnas_ordenadas]

# Guardar archivo CSV
ruta_completa_salida = os.path.join(ruta_salida, nombre_archivo_salida)
promedios.to_csv(ruta_completa_salida, index=False, encoding="utf-8", na_rep="NaN")

print("Promedios calculados correctamente en:", ruta_completa_salida)
```
