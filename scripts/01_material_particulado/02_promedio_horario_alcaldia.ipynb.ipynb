```python
#USAR PARA PM10

import pandas as pd
import numpy as np
import os

# === CONFIGURACIÓN DE ARCHIVOS ===
ruta_archivo = "AGREGAR RUTA DEL ARCHIVO"
ruta_salida = "AGREGAR RUTA DE CARPETA DE SALIDA"
nombre_archivo_salida = "AGREGAR NOMBRE DE SALIDA.csv”

# Crear carpeta si no existe
os.makedirs(ruta_salida, exist_ok=True)

# === CARGAR Y LIMPIAR DATOS ===
df = pd.read_csv(ruta_archivo)

# Convertir FECHA a datetime
df['FECHA'] = pd.to_datetime(df['FECHA'], format="%d/%m/%y", errors='coerce')

# Asegurar que HORA sea numérica entera
df['HORA'] = pd.to_numeric(df['HORA'], errors='coerce').astype('Int64')

# Reemplazar "no data" y variantes por NaN
df.replace(["no data", " no data", "NO DATA", "", "NaN"], np.nan, inplace=True)

# Convertir columnas numéricas (excepto FECHA y HORA)
for col in df.columns:
    if col not in ["FECHA", "HORA"]:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# === DICCIONARIO DE AGRUPACIÓN POR DELEGACIÓN ===
agrupaciones = {
   
    "Alvaro Obregon": ["PED","PLA"],
    "Azcapotzalco": ["CAM"],
    "Benito Juarez": ["BJU"],
    "Coyoacan": ["TAX"],
    "Cuajimalpa de Morelos": ["SFE","CUA"],
    "Cuauhtemoc": ["HGM"],
    "Gustavo A. Madero": ["GAM", "LVI"],
    "Iztalco": ["IZT"],
    "Iztapalapa": ["CES","UIZ", "SAC"],
    "Miguel Hidalgo": ["LPR","MGH"],
    "Milpa Alta": ["MPA"],
    "Tlahuac": ["TAH"],
    "Tlalpan": ["AJM"],
    "Venustiano Carranza": ["HAN","MER"],
    "Acolman": ["ACO"],
    "Atizapan de Zaragoza": ["ATI"],
    "Chalco": ["CHO"],
    "Coacalco de Berriozabal": ["VIF"],
    "Ecatepec de Morelos": ["LLA","SAG", "XAL"],
    "Naucalpan de Juarez": ["FAC"],
    "Nezahualcoyotl": ["FAR", "NET"],
    "Ocoyoacac": ["INN"],
    "Tepotzotlan": ["CUT"],
    "Tlalnepantla de Baz": ["TLA"],
    "Tultitlan": ["TLI"]
}

# === CALCULAR PROMEDIOS CON REDONDEO ===
df_resultado = pd.DataFrame()
df_resultado['FECHA'] = df['FECHA']
df_resultado['HORA'] = df['HORA']

# Orden alfabético de delegaciones
for delegacion in sorted(agrupaciones.keys()):
    estaciones = agrupaciones[delegacion]
    estaciones_validas = [est for est in estaciones if est in df.columns]

    if len(estaciones_validas) == 1:
        df_resultado[delegacion] = df[estaciones_validas[0]]
    elif len(estaciones_validas) > 1:
        promedio = df[estaciones_validas].mean(axis=1, skipna=True)
        # Aplicar redondeo condicional (redondeo simétrico)
        promedio_redondeado = promedio.apply(
            lambda x: np.floor(x) if not np.isnan(x) and x % 1 < 0.5 else
                      np.ceil(x) if not np.isnan(x) else np.nan
        )
        df_resultado[delegacion] = promedio_redondeado

# === REORDENAR COLUMNAS: FECHA, HORA, DELEGACIONES ===
columnas_ordenadas = ['FECHA', 'HORA'] + sorted(agrupaciones.keys())
df_resultado = df_resultado[columnas_ordenadas]

# === GUARDAR RESULTADO ===
archivo_salida = os.path.join(ruta_salida, nombre_archivo_salida)
df_resultado.to_csv(archivo_salida, index=False, na_rep="NaN")

print("Archivo con redondeo aplicado generado exitosamente:", archivo_salida)
```
