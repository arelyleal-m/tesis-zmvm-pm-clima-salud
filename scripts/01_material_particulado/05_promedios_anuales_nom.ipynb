```python
# ========== IMPORTAR LIBRERÍAS ==========
import pandas as pd
import numpy as np
import os

# ========== CONFIGURACIÓN ==========
ruta_archivo = "AGREGAR RUTA DEL ARCHIVO"
ruta_salida = "AGREGAR RUTA DE CARPETA DE SALIDA"
nombre_archivo_salida = "AGREGAR NOMBRE DE SALIDA.csv”

# ========== CARGAR ARCHIVO ==========
df = pd.read_csv(ruta_archivo)

# Renombrar columna 'AÑO' para facilitar operaciones
df.rename(columns={"AÑO": "ANO"}, inplace=True)

# Asegurarse de que 'ANO' esté como entero
df['ANO'] = pd.to_numeric(df['ANO'], errors='coerce', downcast='integer')

# ========== COLUMNAS A INCLUIR ==========
columnas_excluir = ["ANO", "TRIMESTRE"]
columnas_datos = [col for col in df.columns if col not in columnas_excluir]

# ========== REDONDEO PERSONALIZADO A ENTERO ==========
def redondeo_NOM(valor):
    if pd.isna(valor):
        return np.nan
    primer_decimal = int((valor * 10) % 10)
    return np.floor(valor) if primer_decimal <= 4 else np.ceil(valor)

# ========== FUNCIONES AUXILIARES ==========
def calcular_promedio_anual(grupo):
    resultados = {}
    for col in columnas_datos:
        datos_validos = pd.to_numeric(grupo[col], errors='coerce').dropna()
        if len(datos_validos) >= 3:  # al menos 3 trimestres válidos
            promedio = datos_validos.sum() / len(datos_validos)
            resultados[col] = redondeo_NOM(promedio)
        else:
            resultados[col] = np.nan
    return pd.Series(resultados)

# ========== AGRUPAR Y CALCULAR PROMEDIOS ANUALES ==========
promedios_anuales = df.groupby("ANO").apply(calcular_promedio_anual).reset_index()

# ========== GUARDAR RESULTADOS ==========
salida_completa = os.path.join(ruta_salida, nombre_archivo_salida)
promedios_anuales.to_csv(salida_completa, index=False, encoding="utf-8", na_rep="NaN")

print(f"Promedios anuales calculados correctamente en: {salida_completa}")
```
