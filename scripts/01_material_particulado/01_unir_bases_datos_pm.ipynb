```python
#Importar librerias
import pandas as pd
import os

# Carpeta donde están los archivos CSV
carpeta = 'CARPETA'

# Lista de columnas de interés
columnas_interes = ["NOMBRE COLUMNAS DE INTERES"]

# Lista para almacenar los DataFrames
dfs = []

# Leer y combinar los archivos
for archivo in sorted(os.listdir(carpeta)):  
    if archivo.endswith(".csv"):
        ruta_completa = os.path.join(carpeta, archivo)
        df = pd.read_csv(ruta_completa)

        # Filtrar solo las columnas de interés que existan en el archivo
        columnas_existentes = [col for col in columnas_interes if col in df.columns]
        df = df[columnas_existentes]

        dfs.append(df)

# Unir todos los DataFrames, respetando las columnas de interés
df_final = pd.concat(dfs, ignore_index=True)

# Llenar los espacios vacíos con "no data"
df_final.replace(-99, "no data", inplace=True)  # Reemplazar valores -99
df_final = df_final.fillna("no data")  # Reemplazar valores vacíos

# Reordenar columnas: "FECHA" primero, "HORA" segundo, el resto en orden alfabético
columnas_ordenadas = ['FECHA', 'HORA'] + sorted([col for col in df_final.columns if col not in ['FECHA', 'HORA']])
df_final = df_final[columnas_ordenadas]

# Guardar el resultado
df_final.to_csv("NOMBRE DE SALIDA.csv", index=False)

print("Archivos combinados exitosamente")

# Mostrar los primeros resultados
print(df_final.head())
```
